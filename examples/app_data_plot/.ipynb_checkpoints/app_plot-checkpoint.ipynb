```python
topo_filename = "./topo.txt"
data_filename = "./plt1.txt"
```

```python
topo_xyz = np.loadtxt(str(topo_filename))
dc_data = read_dcip2d_ubc(data_filename, 'volt', 'surface')
```

```python
from palettable.lightbartlein.diverging import BlueDarkRed18_18
fixed_cmap = BlueDarkRed18_18.mpl_colormap
# Plot voltages pseudo-section

import pickle
with open('cbar_params.pickle', 'rb') as file:
    cbar_params = pickle.load(file)
    
sm = plt.cm.ScalarMappable(cmap=cbar_params['cmap'], norm=cbar_params['norm'])  


fig = plt.figure(figsize=(12, 5))


ax1 = fig.add_subplot(1,1,1)
ax1, cc,cbar = plot_pseudosection(
    dc_data,
    plot_type="scatter",
    ax=ax1,
    scale="linear",
    cbar_label="Apparent resistivity ($\Omega$ m)",
    mask_topography=True,
    pcolor_opts={"cmap": fixed_cmap},
    contourf_opts={"cmap": fixed_cmap},
    scatter_opts={"cmap": fixed_cmap,'marker':'s','s':10},
    data_locations=False,
    cbar_opts={'location': 'right','orientation': 'vertical','format':"%.0f",
              'fraction':0.01,'shrink':0.8,'aspect':30},
    clim=[100,1000]
)
cbar = plt.colorbar(sm, ax=ax1)
# cbar.clim([60,600]) 
ax1.set_ylabel(' ')
ax1.set_xlabel(' ')
ax1.spines['top'].set_visible(False)
ax1.spines['right'].set_visible(False)
ax1.spines['bottom'].set_visible(False)
ax1.spines['left'].set_visible(False)
ax1.get_xaxis().set_ticks([])
ax1.get_yaxis().set_ticks([]) 
#cbar.remove()
# y_ticks = np.linspace(0,20, 5) 
# ax1.set_yticks([9999-y_ticks[0],9999-y_ticks[1],9999-y_ticks[2]
# ,9999-y_ticks[3],9999-y_ticks[4]])
# ax1.set_yticklabels(y_ticks)
#ax1.set_xlim([0,70])
#fig.savefig('ERT3.tiff',dpi=300,tight_layout=True)
```

```python
fig = plt.figure(figsize=[12,6])
nn = 1

for i in range(12):
    dc_data = read_dcip2d_ubc('Plt'+str(i)+'.txt', 'volt', 'surface')
    ax1 = fig.add_subplot(3,4,nn)
    nn = nn +1 
    ax1, cc,cbar = plot_pseudosection(
        dc_data,
        plot_type="pcolor",
        ax=ax1,
        scale="linear",
        cbar_label="Apparent resistivity ($\Omega$ m)",
        mask_topography=True,
        pcolor_opts={"cmap": fixed_cmap},
        data_locations=False,
        cbar_opts={'location': 'right','orientation': 'vertical','format':"%.0f",
                  'fraction':0.10,'shrink':0.8,'aspect':30},
        clim=[100,1000]
    )
    #cbar = plt.colorbar(sm, ax=ax1)
    # cbar.clim([60,600]) 
    ax1.set_ylabel(' ')
    ax1.set_xlabel(' ')
    ax1.spines['top'].set_visible(False)
    ax1.spines['right'].set_visible(False)
    ax1.spines['bottom'].set_visible(False)
    ax1.spines['left'].set_visible(False)
    ax1.get_xaxis().set_ticks([])
    ax1.get_yaxis().set_ticks([]) 
    cbar.remove()
plt.tight_layout()
fig.savefig('Monthly_appRes.tif',bbox_inches='tight',dpi=100)
```
