```python
import os
import numpy as np
import matplotlib.pyplot as plt
import numpy as np
import pygimli as pg
from pygimli.physics import ert
data_dir = './'
files = os.listdir(data_dir)


ertfile = []
Data_arr = []
for i in np.arange(30,365,30) + 365*2:

    ertfile.append('ertdata' + str(i))

    Data_arr.append(float(i))
                
index = np.argsort(Data_arr)
new_Date_arr = np.sort(Data_arr)
ertfile = np.array(ertfile)
new_ertfile = ertfile[index]

###########################################################################################

#name = new_ertfile[range(365)[180::30]]  # the file name array for ERT data file in right time order
#Date = new_Date_arr[range(365)[180::30]] # the time array for corresponding ERT data file, in np.datetime64 format

name = new_ertfile # the file name array for ERT data file in right time order
Date = new_Date_arr   # the time array for corresponding ERT data file, in np.datetime64 format
```

```python
def data_process(filename,outfilename):
    ertData = ert.load(filename)
    #ertData['k'] = ert.createGeometricFactors(ertData, numerical=True)
    #k0 = ert.createGeometricFactors(ertData)
    mgr = ert.ERTManager(ertData)

    #ertData['rhoa'] = ertData['k']*ertData['u']/ertData['i']
    
    xx = np.array(pg.x(ertData.sensorPositions()))
    yy = np.array(pg.y(ertData.sensorPositions()))
    from pandas import Series,DataFrame
    import pandas as pd
    
    ss = {'a':ertData['a'],
      'b':ertData['b'],
      'm':ertData['m'],
      'n':ertData['n'],
      'u':ertData['u'],
      'i':ertData['i'],
      'rhoa':ertData['rhoa'],
      'err':ertData['err']}
    
    ss = DataFrame(ss)
    res = ss
    
    print("! Dc data", file=open(outfilename, "w"))
    print("COMMON_CURRENT", file=open(outfilename, "a"))
    print("1", file=open(outfilename, "a"))
    for i in range(res.shape[0]):
        print(str(xx[res['a'][i]])+' '+str(yy[res['a'][i]])+' '
              +str(xx[res['b'][i]])+' '+str(yy[res['b'][i]])+'  '+'1', 
              file=open(outfilename, "a"))
        print(str(xx[res['m'][i]])+' '+str(yy[res['m'][i]])+' '
          +str(xx[res['n'][i]])+' '+str(yy[res['n'][i]])+'  '+str(res['rhoa'][i])+'  '+str(res['err'][i]), 
          file=open(outfilename, "a"))
        
    xyz = np.vstack((xx,np.zeros((yy.shape)),yy))
    np.savetxt('topo.txt',xyz.T)
```

```python
for i in range(len(name)):
    data_process(name[i],'Plt'+str(i)+'.txt')
```
