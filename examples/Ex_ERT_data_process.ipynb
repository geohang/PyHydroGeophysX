# ERT data process

This notebook demonstrates the ERT data process workflow in PyHydroGeophysX.


```python
import os
import sys
import importlib

# Setup package path for development
try:
    # For regular Python scripts
    current_dir = os.path.dirname(os.path.abspath(__file__))
except NameError:
    # For Jupyter notebooks
    current_dir = os.getcwd()

# Add the parent directory to Python path
parent_dir = os.path.dirname(current_dir)
if parent_dir not in sys.path:
    sys.path.append(parent_dir)

# Force reload of the module to pick up recent changes
import PyHydroGeophysX.data_processing.ert_data_agent as ert_agent
importlib.reload(ert_agent)
    
from PyHydroGeophysX.data_processing.ert_data_agent import (
    load_ert_resipy, qc_and_visualize, export_for_inversion, LocalRef
)

ert = load_ert_resipy(
    project_dir="data/ERT/E4D",
    data_file="data/ERT/E4D/2021-10-08_1400.ohm",  # Full relative path from current directory
    instrument="E4D",      # explicit instrument
    crs="local",              # default; use crs="EPSG:32615", epsg=32615 for projected coords
    local_ref=LocalRef(origin_x=0.0, origin_y=0.0, azimuth_deg=90.0)
)
artifacts = qc_and_visualize(ert, outdir="results/ert_data_process")  # Changed from /results to results
bert_path = export_for_inversion(ert, outdir="results/ert_data_process", fmt="pgimli")
print("Artifacts:", artifacts)
print("BERT export:", bert_path)
```
